```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
class_count = titanic['Pclass'].value_counts().sort_index()
class_label = ['1st class', '2nd class', '3rd class']

plt.figure(figsize=(6, 6))
plt.pie(class_count,
        labels=class_label,
        colors=['red', 'green', 'blue'],
        autopct="%0.2f%%",
        wedgeprops={'width': 0.5})
plt.title('Passenger distribution by class', loc='center')
plt.show()
```

# 2. Count plot

```python
plt.figure(figsize=(5, 6))
sns.countplot(data=titanic, x='Embarked')
plt.title('Number of passengers by embarkation port')
plt.xlabel('Embarked')
plt.ylabel('Count')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(data = iris, x = 'sepal_length')
sns.rugplot(data = iris, x = 'sepal_length')
```

# 4. Violin plot

```python
sns.violinplot(data = titanic, y = 'Pclass', x = 'Survived', palette = "pastel")
```

# 5. Box plot

```python
sns.boxplot(data = titanic, y = 'Age', x = 'Survived', palette = 'Set2')
```

# 6. Boxen plot

```python
sns.boxenplot(data = titanic, x = 'Fare')
```

# 7. Histogram

```python
data = titanic['Age'].dropna()

plt.figure(figsize = (8, 5))
plt.hist(data,
        bins = 15,
        color = 'g',
        alpha = 0.75,
        edgecolor = 'r',
        cumulative = False)

plt.title('Age histplot', loc = 'center')
plt.xlabel('Age')
plt.ylabel('Number of passenger')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Total bill vs Tip')
plt.show()
```

# 2. Line plot

```python
avg_fare_by_class = titanic.groupby('Pclass')['Fare'].mean()

plt.figure(figsize=(8, 5))
plt.plot(avg_fare_by_class.index, avg_fare_by_class.values,
         color='b', marker='o', markersize=10, linewidth=2)
plt.title('Average fare by passenger class')
plt.xlabel('Pclass')
plt.ylabel('Average Fare')
plt.xticks([1, 2, 3, 4])
plt.grid(color='grey', linestyle='--', alpha=0.5)
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
plt.show()
```

# 4. Hexbin plot

```python
plt.figure(figsize=(8, 5))
plt.hexbin(tips['total_bill'], tips['tip'], gridsize=20, cmap='Blues')
plt.colorbar(label='count')
plt.title('Total bill vs Tip (Hexbin plot)')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

# 5. 2D kde plot

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=tips, x='total_bill', y='tip', cmap='Reds', fill=True)
plt.title('Total bill vs Tip (2D KDE plot)')
plt.show()
```

# 6. Box plot

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total bill by day')
plt.show()
```

# 7. Violin plot

```python
plt.figure(figsize=(8, 2))
sns.violinplot(data=tips, x='day', y='tip')
plt.title('Tip distribution by day')
plt.show()
```

# 8. Swarm plot

```python
plt.figure(figsize=(8, 5))
sns.swarmplot(data=tips, x='day', y='total_bill')
plt.title('Total bill by day (Swarm plot)')
plt.show()
```

# 9. Count plot

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=titanic, x='Pclass', hue='Survived')
plt.title('Survival count by passenger class')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='day', size='size', style='sex')
plt.title('Total bill vs Tip by day, size and sex')
plt.show()
```

# 2. Pair plot

```python
g = sns.PairGrid(iris, hue = 'species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)

plt.show()
```

# 3. Heatmap

```python
sns.heatmap(titanic.select_dtypes(include = 'number').corr(), cmap = 'coolwarm', annot = True)
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(tips, col = 'day', row = 'time')
g.map(sns.histplot, 'total_bill')
plt.show
```

# 5. PairGrid

```python
g = sns.PairGrid(iris, hue = 'species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)

plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data=tips, x='total_bill', y='tip', hue='day', kind='scatter')
plt.show()
```

# 7. Boxen plot

```python
plt.figure(figsize=(8, 3))
sns.boxenplot(data=tips, x='day', y='total_bill', hue='time')
plt.title('Total bill by day and time (Boxen plot)')
plt.show()
```
